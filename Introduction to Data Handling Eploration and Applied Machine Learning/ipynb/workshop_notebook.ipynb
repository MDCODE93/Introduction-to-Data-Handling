# Session 2 Workshop — Git & GitHub Practice

This notebook is the file you'll download from Moodle, edit, and push to your own GitHub repo during today's hands-on.

Don't overthink the Python — the point of this notebook is practicing the loop: **edit &rarr; `git add` &rarr; `git commit` &rarr; `git push`**.

## Step 1 — Set up your own repo

Save this notebook (downloaded from Moodle) into a new folder on your laptop, then:

1. On **github.com**: New repository → name it → Public or Private → Create (leave "Add a README" unchecked — start empty)
2. Open a terminal **in that folder** and run:

```bash
git init
git remote add origin https://github.com/your-username/your-repo.git
git status
```

You should see this notebook listed as untracked.

````python
name = "Alex"                 # a variable
print(f"Hello, {name}!")      # prints text

for n in [1, 2, 3]:           # runs the line below 3 times
    print(n)
````

## Step 2 — Make it yours

Edit the `name` variable above to your own name. Run the cell (**Shift+Enter**) and check the output changed.

## Step 3 — Save & push it

Save the notebook (**Ctrl/Cmd+S**), then back in the terminal:

```bash
git status
git add workshop_notebook.ipynb
git commit -m "personalize hello message"
git push -u origin main
```

The `-u` connects your local `main` branch to GitHub's — you only need it on this first push.

## Step 4 — Verify

Open your repo on **github.com** and confirm:
- your commit shows up in the repo's history
- the notebook file itself shows your edited name
